```python
sns.scatterplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex')
sns.despine()
```

```python
sns.scatterplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex')
sns.despine()
sns.set_context('talk')
# we have talk, notebook(default), poster and paper(small) contexts 
```

```python
sns.scatterplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex', palette = 'Set1')
sns.despine()
```

```python
sns.scatterplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', hue = 'sex', palette = 'Set1', style = 'sex')
sns.despine()
# now we will classify the data on the basis of sex, (using in style = 'xyz') means we give different signs to all those in the group
```

```python
sns.regplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g')
# we use regplot instead of scatterplot when we want to draw a straight line passing through most points
```

```python
sns.regplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', scatter=False)
sns.despine()
#  we can also make the data disappear
```

```python
# we also have a line plot displaying variations in sort of a frequency tyoe of graph
sns.lineplot(data=iris, x = 'flipper_length_mm', y = 'body_mass_g', hue='sex')
```

```python
# we'll now plot bar graphs for the data
sns.barplot(data=iris, x = 'species', y = 'body_mass_g', hue = 'sex')
sns.despine()
```

```python
# we can also add various types of scales like based on mean or avg, median, mode, etc...
sns.barplot(data=iris, x = 'species', y = 'body_mass_g', hue = 'sex', estimator = np.sum)
sns.despine()
```

```python
sns.barplot(data=iris, x = 'species', y = 'body_mass_g', hue = 'sex', estimator = np.mean) # mean is by deafult
sns.despine()
```

```python
# we can also find variance
sns.barplot(data=iris, x = 'species', y = 'body_mass_g', hue = 'sex', estimator = np.var)
sns.despine()
```

```python
# we can also find the number of items with a specific label
# this is called countplot

sns.countplot(data=iris, x = 'species')
```

```python
# we also have box plots which area used to detect outliers in the data
sns.boxplot(data=iris, x = 'species', y = 'body_mass_g', hue = 'sex')
# the data below shows that one male and one female species have extreme high and extermely low body mass compared to the others
```

```python
# we use headmaps to see the coorelation of a data to another using numbers, note that the data should be a number too for this to work
iris.head()
```

```python
column = ['bill_length_mm' 	,'bill_depth_mm' 	,'flipper_length_mm' 	,'body_mass_g']
iris[column].corr()
# here we can see how much a data affects the other, now we plot this for easier understanding
```

```python
sns.heatmap(data = iris[column].corr())
```

```python
# we can also annotate this with the numbers in the blocks for better understanding
sns.heatmap(data = iris[column].corr(), annot = True)
# light colors mean more connection betweent the two datas
```
